```python
import asyncio

async def task_one():
    await asyncio.sleep(2)
    print("Task One Complete")

async def task_two():
    await asyncio.sleep(1)
    print("Task Two Complete")

async def main():
    t1 = asyncio.create_task(task_one())
    t2 = asyncio.create_task(task_two())

    print("Both tasks started...")
    await t1
    await t2
    print("All tasks done!")

await main()
```

```python
import asyncio

async def task_one():
    await asyncio.sleep(2)
    print("Task One Complete")

async def task_two():
    await asyncio.sleep(1)
    print("Task Two Complete")

async def main():
    print("Both tasks started...")
    await asyncio.gather(task_one(), task_two())
    print("All tasks done")

await main()
```

```python
import asyncio

async def start_task(number: str, sleep: int):
    await asyncio.sleep(sleep)
    print(f"Task {number.capitalize()} Complete")

async def main():
    print("Both tasks started...")
    await asyncio.gather(start_task("One", 2), start_task("Two", 1))
    print("All tasks done")

await main()
```
