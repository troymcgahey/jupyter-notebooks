## 1. Visualizing a vector

A 2-D vector

\[
x = [2, 1]
\]

can be visualized as an arrow from the origin `(0, 0)` to the point `(2, 1)`.


```python
x = np.array([2, 1])

fig, ax = plt.subplots()

ax.quiver(
    0, 0,
    x[0], x[1],
    angles="xy",
    scale_units="xy",
    scale=1,
)

ax.axhline(0)
ax.axvline(0)
ax.grid()
ax.set_aspect("equal")
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)

plt.show()
```

## 2. Dot product

For two vectors:

\[
a \cdot b = \|a\|\|b\|\cos(\theta)
\]

The dot product is a scalar that combines the vectors' magnitudes with their directional alignment.


```python
a = np.array([2, 1])
b = np.array([1, 3])

dot_product = np.dot(a, b)

print("a =", a)
print("b =", b)
print("a · b =", dot_product)
```

### Exercise 2 — Angle between vectors

Use the relationship

\[
\cos(\theta) = \frac{a \cdot b}{\|a\|\|b\|}
\]

to calculate the angle between `a` and `b`.

Helpful NumPy functions:

- `np.linalg.norm()`
- `np.arccos()`
- `np.degrees()`


```python
# TODO: Calculate the magnitude of a
a = np.array([2, 1])
b = np.array([1, 3])

magnitude_a = np.linalg.norm(a)
print("Length of A:", magnitude_a)

# TODO: Calculate the magnitude of b
magnitude_b = np.linalg.norm(b)
print("Length of B:", magnitude_b)

dot_product = np.dot(a, b)
print("Dot product:", dot_product)

# TODO: Calculate cos(theta)
cos_theta = dot_product/((magnitude_a)*(magnitude_b))
print("Cosine theta:", cos_theta)

# TODO: Calculate theta in degrees
theta = np.arccos(cos_theta)
print("Theta:", theta)

theta_degress = np.degrees(theta)
print("Angle:", theta_degress)
```

## 3. Matrix transformation

Now we'll transform a row vector using a matrix.

\[
x = [2,1]
\]

and

\[
A =
\begin{bmatrix}
2 & 0 \\
0 & 1
\end{bmatrix}
\]

Using our row-vector convention:

\[
x' = xA
\]

This matrix stretches the x-direction by a factor of 2 while leaving the y-direction unchanged.


```python
x = np.array([2, 1])

A = np.array([
    [2, 0],
    [0, 1]
])

x_transformed = x @ A

print("Original:", x)
print("Transformed:", x_transformed)
```

```python
fig, ax = plt.subplots()

ax.quiver(
    0, 0,
    x[0], x[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="Original"
)

ax.quiver(
    0, 0,
    x_transformed[0], x_transformed[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="Transformed"
)

ax.axhline(0)
ax.axvline(0)
ax.grid()
ax.set_aspect("equal")
ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.legend()

plt.show()
```

## 4. Transforming the basis vectors

The standard basis vectors are:

\[
e_1 = [1,0]
\]

\[
e_2 = [0,1]
\]

Watching where a matrix sends these two vectors gives us a powerful way to understand what the matrix does to the entire coordinate system.


```python
e1 = np.array([1, 0])
e2 = np.array([0, 1])

A = np.array([
    [2, 1],
    [0, 1]
])

e1_transformed = e1 @ A
e2_transformed = e2 @ A

print("e1 ->", e1_transformed)
print("e2 ->", e2_transformed)
```

## 5. Linear dependence and dimension collapse

Consider:

\[
A =
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}
\]

The second row is twice the first, so the rows are linearly dependent.

Let's transform many vectors and see what happens.


```python
A_dependent = np.array([
    [1, 2],
    [2, 4]
])

vectors = np.array([
    [1, 0],
    [0, 1],
    [1, 1],
    [2, -1],
    [-2, 2],
    [3, 1],
])

transformed = vectors @ A_dependent

print(transformed)
```
